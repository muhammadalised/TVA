# OnHW handwriting tokenizer — historical IAM+READ condition

This prepares the frozen 419-class vocabulary used by the earlier OnHW experiments. For the current IAM-only ED comparison, use train_ed_tokenizers.ipynb.


```python
from pathlib import Path
from tva.handwriting_tokenizers import HandwritingBigramTokenizer

source = Path('/home/artellisys/dtlr-output/combined-iam-read-v1/model.json')
output = Path('artifacts/tokenizers/onhw_words500_rh_iam_read_v1.json')
categories = ['', *list('ABCDEFGHIJKLMNOPQRSTUVWXYZÄÖÜabcdefghijklmnopqrstuvwxyzäöüß')]
```

```python
tokenizer = HandwritingBigramTokenizer()
tokenizer.train(source, categories, output_path=output)
print('Classes:', tokenizer.size)
```

```python
text = 'Dabei'
print(tokenizer.segment(text)['tokens'])
print(tokenizer.decode(tokenizer.encode(text)))
```
